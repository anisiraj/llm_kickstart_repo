# Case Study §10 — Deploy with Ollama (Part B)

Runnable twin of [`10_ollama_deploy.py`](10_ollama_deploy.py). Ollama imports the model (it can read
Llama-arch **safetensors** directly, so it works on the §9 merged model even without a GGUF), serves it
on a local REST API, and we query it — measuring **local tokens/sec**, the metric that matters at the edge.

Gracefully skips if the `ollama` binary/server isn't available. In TRIAL the fine-tuned 135M is barely
trained and emits EOS early (empty answers, tok/s N/A) — the **deployment** still proves out; real
answers + a real tok/s come from a FULL fine-tune.

```python
MODE = "trial"
import importlib.util, pathlib, sys
HERE = pathlib.Path.cwd()
root = HERE if (HERE / "10_ollama_deploy.py").exists() else HERE / "case_study"
sys.path.insert(0, str(root))
import config; config.set_mode(MODE)
spec = importlib.util.spec_from_file_location("s10", root / "10_ollama_deploy.py")
s10 = importlib.util.module_from_spec(spec); spec.loader.exec_module(s10)
print('ollama server reachable:', s10._server_up())
```

```python
info = s10.run()
print(info)
```

```python
if info.get('deployed'):
    print('\u2713 §10 verified: model imported into Ollama and served over its local API.')
else:
    print('\u2139 §10 skipped:', info.get('skipped'), '\u2014 install/start Ollama to run this section.')
print('Next: \u00a711 edge benchmark (tok/s + RAM on a Raspberry Pi-class device).')
```
