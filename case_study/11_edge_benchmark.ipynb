# Case Study §11 — Edge benchmark: footprint + speed

Runnable twin of [`11_edge_benchmark.py`](11_edge_benchmark.py). The payoff of a tiny model is on-device
inference, so we measure the deployed model's **footprint** (quantized size on disk ≈ RAM proxy) and
**speed** (local tokens/sec) via Ollama.

**To reproduce on a Raspberry Pi:** install Ollama there, `ollama create` the same Modelfile, hit the
same `/api/generate` — the code is identical, only the hardware changes. Reference (RESEARCH_NOTES §5):
Pi 5 runs 1–3B at ~10–18 tok/s; a 135M model is far lighter and faster.

```python
MODE = "trial"
import importlib.util, pathlib, sys
HERE = pathlib.Path.cwd()
root = HERE if (HERE / "11_edge_benchmark.py").exists() else HERE / "case_study"
sys.path.insert(0, str(root))
import config; config.set_mode(MODE)
spec = importlib.util.spec_from_file_location("s11", root / "11_edge_benchmark.py")
s11 = importlib.util.module_from_spec(spec); spec.loader.exec_module(s11)
m = s11.run()
```

```python
if not m.get('skipped'):
    print(f"footprint: {m['footprint']['size_gb']:.3f} GB | params {m['footprint']['params']} | quant {m['footprint']['quant']}")
    print('speed:', f"{m['speed']['tok_per_s']:.0f} tok/s" if m['speed']['tok_per_s'] else 'N/A')
    assert m['footprint']['size_gb'] > 0
    print('\u2713 §11 verified: on-device footprint + speed measured (reproduce on a Pi with the same code).')
else:
    print('\u2139 §11 skipped:', m['skipped'])
print('Next: \u00a712 \u2014 lm-evaluation-harness + run-as-agent.')
```
